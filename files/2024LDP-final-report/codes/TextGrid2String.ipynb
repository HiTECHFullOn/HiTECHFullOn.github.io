```python
vowel_intervals = []    #모든 모음의 발화구간에 대한 정보 저장
for f in tqdm(files):
    text1 = tg.TextGrid.fromFile(f'label-TextGrid/{f}') #TextGrid 클래스로 불러오기
    intervals = [f, ]
    for i, t in enumerate(text1[0]):    #첫 번째 tier인 phoneme tier 안의 각 구간에 대해
        start = t.minTime               #구간 시작 시간,
        end = t.maxTime                 #구간 끝 시간,
        content = t.mark                #구간 내용을 변수로 저장
        if content[0]!='<':             #휴지나 웃음소리 등 특수 토큰이 아니면
            phone = seoul2ipa[content]  #해당 음소를 IPA로 변환
        else: phone = '#'
        if phone in V:
            if i == 0:                  #첫 구간이면 앞은 경계, 뒤는 음소 취급
                fore = '#'
                back = seoul2ipa[text1[0][i+1].mark] if text1[0][i+1].mark in seoul2ipa.keys() else '#'
            elif i == (len(text1[0])-1):#마지막 구간이면 그 뒤는 경계, 앞은 음소
                fore = seoul2ipa[text1[0][i-1].mark] if text1[0][i-1].mark in seoul2ipa.keys() else '#'
                back = '#'
            else:                       #이외에는 앞뒤 모두 음소 취급, 환경 탐색
                fore = seoul2ipa[text1[0][i-1].mark] if text1[0][i-1].mark in seoul2ipa.keys() else '#'
                back = seoul2ipa[text1[0][i+1].mark] if text1[0][i+1].mark in seoul2ipa.keys() else '#'
            intervals.append(((start, end), fore, phone, back)) #앞뒤 음소와 모음 실현 구간을 intervals에 저장
        else: pass
    vowel_intervals.append(intervals)

    vowel_intervals.sort()
```

```python
#각 모음구간에 관한 정보를 파일명을 key로 삼는 dictionary로, 구간 정보는 길이로 변환
seg_time = dict()
for text in vowel_intervals:
    file_name = text[0].split('.')[0]
    lengths = [end-start for start, end in list(zip(*text[1:]))[0]]
    seg = list(zip(*(list(zip(*text[1:]))[1:]), lengths))
    seg_time[file_name] = seg
```

```python
#모음구간의 시작-끝 시간 정보를 분리하여 파일별로 저장; 이는 Praat script의 입력이 될 .String 파일을 생성하기 위함
#모음구간 환경도 분리하여 파일별로 저장; 추후 devoicing rate와 결합하여 DataFrame으로 변환
intervals_only = []
envr = dict()
for text in vowel_intervals:
    file_name = text[0]
    intervals = list(zip(*text[1:]))[0]
    intervals_only.append(intervals)
    environments = list(zip(*text[1:]))[1:]
    envr[file_name] = list(zip(*environments))
```

```python
#모음구간의 리스트를 받아 Praat String 파일로 변환
def itv2str(interval):
    file_name = 'intervals/'+interval[0].split('.')[0]+'.Strings'
    f = open(file_name, 'w')
    numberOfStrings = len(interval)-1
    root = f"""File type = \"ooTextFile\"
Object class = \"Strings\"

numberOfStrings = {numberOfStrings} 
strings []: 
"""
    for i, inter in enumerate(interval[1:]):
        root += f"\tstrings [{i+1}] = "+"\""+str(inter[0][0])+"&"+str(inter[0][1])+"\"\n"
    
    f.write(root)
    f.close()
```

```python
for VI in vowel_intervals:
    itv2str(VI)
#이 과정을 모두 거치면 'intervals' 폴더에 각 TextGrid에 대응되는 모음구간들을 지정해 주는 동 파일명의 .String 파일이 생성됨.
```

```python
#파일 이름 리스트 생성
filelist = sorted(list(seg_time.keys()))
```

```python
#각 파일에 대응하는 무성음화 비율 리스트 생성, 딕셔너리 형태로 저장
devoicing_rate = dict()
for file in filelist[:60]:
    f = open(f'devoicing/{file}'+'.txt')
    processed_lines = list(map(float, list(map(str.strip, f.readlines()))))
    devoicing_rate[file] = processed_lines
```

```python
#모음구간 정보에 무성음화 비율 결합
vowel_sets = {file: [i + (j,) for i, j in (zip(seg_time[file], devoicing_rate[file]))] for file in filelist[:60]}
```

```python
vowel_sets['s01m16f1']
```

```python
#파일별 정보 병합
raw_data = pd.concat([pd.DataFrame(vowel_sets[file]) for file in filelist[:60]], axis=0)
```

```python
#devoicing rate 0.5를 기준으로 binary한 무성음화 여부 할당 및 열 이름 재지정
data = raw_data
data.columns = ['pre', 'V', 'post', 'length', 'devoicing']
data['devoiced'] = data['devoicing']>=0.5
data
```

```python
#자음 간 환경으로 제한
env_exclusion = V.union({'ja', 'je', 'jo', 'ju', 'jə', 'wa', 'we', 'wi', 'wə', '#', 'ɨi'})
filtered_data = pd.concat([data.iloc[i] for i in range(len(data)) if data.iloc[i]['pre'] not in env_exclusion and data.iloc[i]['post'] not in env_exclusion], axis=1).transpose()
filtered_data
```

```python
#csv 포맷으로 저장
filtered_data.to_csv('vowel60.csv', encoding='utf-16', index=False)
```
